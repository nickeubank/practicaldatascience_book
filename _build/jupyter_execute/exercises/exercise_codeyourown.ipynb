### Exercise 3 

When we are writing code to fit a statistical model, it usually makes sense to start by constructing data that has properties you understand. This give you way to validate your model as you go along.

In a sense, this is an inversion of how we usually approach data science — usually, we have data, and we use a model we understand to try and understand the underlying structure of the data. But here, we want to *construct* data whose structure we understand so that we know what our model *should* generate.

To illustrate, here's an example of a toy dataset — an `X` matrix and a `y` vector. The values in `X` are random, but I've carefully specified how `y` should be created. In particular, I've said that `y` should be the result of multiplying the values of `X` by a specified set of coefficients plus a little mean-zero noise.

Because this is precisely how a linear regression tries to decompose relationships in data, a working model should recover the coefficients you started with (or something very close to those coefficients, since you added a little random noise).


```python
import numpy as np
import numpy.random as npr

npr.seed(42)

n = 1_000

# Three columns — one constant and two random variables.
sample_X = np.column_stack([np.ones(n), npr.normal(size=(n, 2))])

# The coefficients I'm using
sample_betas = np.array([2.0, -1.5, 7])

# generate my y
sample_y = sample_X @ sample_betas + npr.normal(scale=1.0, size=n)
```

### Exercise 5 

Now let's make the statisticians proud, and in addition to storing the coefficients, let's store the standard errors for our estimated coefficients as another attribute called `standard_errors` (in the same order as your coefficients). Recall that the simplest method of calculating the variance covariance matrix for $\beta$ is using the formula $\sigma^2 (X'X)^{-1}$, where $\sigma^2$ is the variance of the error terms of your regression. 

The standard errors for your coefficient estimates will be the square root of the diagonal values of that matrix. See page six [here](https://en.wikipedia.org/wiki/Ordinary_least_squares#Matrix/vector_formulation) for a full derivation.

To estimate $\sigma^2$, divide the sum of your squared residuals by $n - k$, where $n$ is the number of observations and $k$ is the number of coefficients you estimated (including the constant term).
